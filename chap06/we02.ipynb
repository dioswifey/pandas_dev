```python
import pandas as pd
```

```python
#데이터 로딩 -> 데이터 프레임 생성
cust = pd.read_csv('../data/customers.csv')
```

```python
#데이터 탐색
cust.head()
```

```python
#도시명에 공백 문제 발생
print(f'정제 전 도시 종류 수 : {cust["city"].nunique()}')

print(f'도시명 :  {cust["city"].unique()}')


#Dataframe: index, values, columns 크게 3개로 구성 -> 이건 3개를 따로따로 뽑을 수 있음 
#.value_counts() 만 하면 series 로 나오고 -> index와 values 중 골라서 뺄수 있음
print(f'도시명 :  {cust["city"].value_counts().index}') #series 에서 index 추출
print(f'도시명 :  {cust["city"].value_counts().values}') #series 에서 values 추출
```

```python
#정제 후 확인

#문자열 함수: strip(), 양쪽 공백 제거, 원본 반영 안됨
#원본 반영안되기 때문에 다시 대입을 해줘야함
cust['city'] = cust['city'].str.strip()
```

```python
print(f'정제 후 도시 종류 수: {cust["city"].nunique()}')
print(f'정제 후 도시: \n {cust["city"].value_counts()}')
```
