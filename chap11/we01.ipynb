#### 데이터 형태

##### categorical (string , int)
- 예 : 도시, 성별, 학과, 요일, 혈액형
- 명목형: 순서없음
- 순서형: 순서있음


##### Countinuous
- 예 : 키, 몸무게, 매출액, 점수

##### 구분법 : 값 두개를 고른 뒤 평균을 낸다. 말이 되면 연속형, 안되면 범주형
