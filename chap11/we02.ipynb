```python
import pandas as pd
import numpy as np
```

```python
#이 코드는 큰 CSV 파일을 한 번에 전부 메모리에 올리지 않고,
#  20만 행씩 나눠 읽으면서 event_type별 개수를 누적 집계하는 코드예요.


total = pd.Series(dtype='int64')
n_chunks = 0


for chunk in pd.read_csv(
    '../data/web_logs.csv',
    usecols=['event_type'],
    chunksize=200000
):
    total = total.add(   #add 누적
        chunk['event_type'].value_counts(),fill_value=0)
    n_chunks += 1

print (f'처리한 청크수 : {n_chunks}')
print(total)



# 처리한 청크수 : 5
# event_type
# cart        119434.0        #5번의 chunk가 add된 누적합
# purchase     80359.0
# search      200206.0
# view        600001.0
# dtype: float64
```

```python
#Parque / csv 비교 :크기, 속도
logs = pd.read_csv('../data/web_logs.csv')
logs['event_type']=logs['event_type'].astype('category')

#Parquet 파일 변환
logs.to_parquet('../data/web_logs.parquet', engine='pyarrow')
```

```python
import time, os
#읽기 속도 비교
#csv 파일 읽고 시간 계산
t0 = time.perf_counter()  #시작 시간 저장
_= pd.read_csv('../data/web_logs.csv')
t1 = time.perf_counter()  #끝난 시간 저장

t_csv = t1 - t0

#parquet 파일 읽는 시간 계산
t0 = time.perf_counter()
_= pd.read_parquet('../data/web_logs.parquet', engine='pyarrow')
t1= time.perf_counter()
t_pq = t1 - t0
print(f'CSV read time:  {t_csv}, PQ read time: {t_pq}')
```
