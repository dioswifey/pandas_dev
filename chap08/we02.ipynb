```python
import pandas as pd
```

```python
#고객의 연령대, 성별 분포(명수) 확인

#데이터 전처리 - drop_duplicates()
# 5000명 -> 4950 명 의 duplicates 가 줄음 
cust = pd.read_csv('../data/customers.csv').drop_duplicates()
cust.info()

#중요한 컬럼: gender, birth_date
#birth_date 가지고 나이 
```

```python
#데이터 전처리
#1. NaN 제거

#4950 =>  4286 
cust = cust.dropna(subset=['gender', 'birth_date']) #.copy()
cust.info()
```

```python
#birth_date 컬럼의 데이터타입이 현재 str => datetime64 변경

cust['birth_date'] = pd.to_datetime(cust['birth_date'])
cust.info()
```

```python

cust.head()
```

```python
#나이 (age) 구하기

#2024년도 기준 나이 구하기
cust['age'] = 2024 - cust['birth_date'].dt.year
cust[['birth_date', 'age']].head()
```

```python
# 나이 구하고 연령대 구하기

#연령대 : '20대 이하', '30-40대', '50대이상'
#bins로 구간 설정
#쳣 사

bins=[0,29,49,200]
labels=['20대 이하', '30-40대', '50대이상']
cust['age_group']=pd.cut(
    x=cust['age'],
    bins=bins,
    labels=labels
)

cust.head()
```

```python
#성별을 한국어로 통일
cust['gender_ko'] = cust['gender'].map(
    {'M':'남','남':'남','F':'여', '여':'여'}
)
cust.head()
```

```python
#고객의 연령대, 성별 분포(명수) 확인
freq = pd.crosstab(cust['age_group'], cust['gender_ko'])
freq
```

```python
#고객의 연령대, 성별 분포(비율) 확인
#crosstab - index

pd.crosstab(
    cust['age_group'],
    cust['gender_ko'],
    normalize='index'  #각 행의 값이 1로 보고 비율을 구하라
).round(2)
```

```python
#crosstab - columns

pd.crosstab(
    cust['age_group'],
    cust['gender_ko'],
    normalize='columns'  #각 행의 값이 1로 보고 비율을 구하라
).round(2)
```
