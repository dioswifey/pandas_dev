```python
import pandas as pd
```

```python
#order item 과 orders과 customers붙히기

#데이터로딩부터
customers = pd.read_csv('../data/customers.csv')
orders = pd.read_csv('../data/orders.csv')
items = pd.read_csv('../data/order_items.csv')
```

```python
# items()에 orders를 병합

items.columns, orders.columns   #이걸 찍어보니 order_id 가 겹친다  => innder join
```

```python
#1 . 주문 상세 order items + 주문 orders: order_id 

#왼쪽이 items, 오른쪽이 orders
oi = items.merge(
    right=orders,
    on='order_id',
    how='inner'
)
items.shape, oi.shape   #1:multiple의 형태가됨
```

```python
# 고객
m = oi.merge(
    right=customers,
    on='customer_id',
    how='inner'
)
m.shape
```

```python
m.info()  # 501199 
#집계전 중복을 잡아야지 안늘어남
```
