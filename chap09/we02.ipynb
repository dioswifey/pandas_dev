```python
customers = pd.read_csv('../data/customers.csv')
products = pd.read_csv('../data/products.csv')
orders =pd.read_csv('../data/orders.csv') #주문
items = pd.read_csv('../data/order_items.csv') #주문 상세
```

```python
#1. 정제
products['price'] = pd.to_numeric(products['price'], errors='coerce')
products['category'] = products['category'].str.strip()
prod = products.drop_duplicates(subset='product_id')
cust = customers.drop_duplicates(subset='customer_id')
items['unit_price'] = pd.to_numeric(items['unit_price'], errors='coerce')
```

```python
prod.info()
```

```python
items.info()
```

```python
#2. 결측 처리

#items 테이블과 pro 테이블
#how = left이니까 왼쪽에 있는 item 기준으로 합친다는 뜻
items = items.merge(
    prod[['product_id', 'price']], on='product_id', how='left'  #50만건
)
items.info()
```

```python
items.info()
```

```python
#NaN처리 : items['unit_price'] = price

#unit price를 채우려고 price를 가져옴 -> price는 이제 필요없어짐
#주문 상세의 단가가 비어있으면, 그 상품의 상품표 기준 가격을 대신 넣겠다라는 로직
# items에는 실제 주문 당시 단가인 unit_price가 있고, prod에는 상품별 기준 가격인 price

#unit_price가 NaN이면 같은 row의 price로 대체
items['unit_price'] = items['unit_price'].fillna(items['price'])
items.info()
```

```python
# price 컬럼 삭제
items = items.drop(columns = 'price')
items.info()
```

```python
# 순차 병합
full = (
    items
    .merge(orders, on = 'order_id', how= 'left', validate='m:1')
    .merge(cust, on='customer_id', how='left', validate='m:1')
    .merge(prod, on='product_id', how='left', validate='m:1')
)
full.head()
```

```python
#매출
full['amount'] = full['unit_price'] * full['quantity'] * (1 - full['discount'])
print(f'통합 테이블 shape: {full.shape}')
print(f'주문 금액 결측 개수: {full["amount"].isna().sum()}')
```

```python
# category(products), 월별(orders), 매출(order_items) 집계 = full DataFrame
full.info()
```

```python
f = full.dropna(subset=['order_datetime'])

f['month'] = f['order_datetime'].dt.month
f.info()
```

```python
#reshaping
#long - > wide  : pivot
# wide -> long

f.groupby(['category', 'month']) #카테고리별 월별 grouping   => DataFrame
f.groupby(['category', 'month'])['amount'] #DataFrame[[컬럼명 multiple]]   매출 컬럼이니까 amount 추출
# f.groupby(['category', 'month',])['amount'].sum()   #multi-index

# rest_index() : category, month index 를 컬럼으로 보내라
# f.groupby(['category', 'month',])['amount'].sum().reset_index()   #인덱스들을 다시 컬럼으로

#원본을 복사본 long으로
long = f.groupby(['category', 'month',])['amount'].sum() #인덱스들을 다시 컬럼으로
#여기서 long은 series

# 튜플 형태로 .loc 조회가 가능한 것은 MultiIndex의 장점 
long.loc[('가구', 4)]
```

```python
#지수 처리하는거
#카테고리 6개 X 월이 6개 => 35개의 row

pd.options.display.float_format='{:,.0f}'.format
long = f.groupby(['category', 'month',])['amount'].sum().reset_index()
long
```

```python
#pivot = 지금 현재 long 상태에서 길게 내려간걸 category랑 month를 reshaping
#pivot에 positional argument / keyword argument방식으로 넣기
#argument  먼저 확인하기, * 들어가있으면 keyword 사용

#positional argument -> error  왜냐하면 *, keyword (처음만 positional, 나머진 keyword)
# wide = long.pivot(
#     'month',
#     'category',
#     'amount'
# )

wide = long.pivot(
    columns='month',
    index='category',
    values='amount'
)
wide
```

```python
#melt : wide -> long
# category를 index 처리에서 빼는 역할은 reset_index()
wide.reset_index()
```

```python
# id_vars='category'는 category를 melt 대상에서 빼고 그대로 유지하라는 뜻이에요.
#category 컬럼은 세로로 녹이지 말고, 각 행을 구분하는 식별자 컬럼으로 유지해라.

wide.reset_index().melt(
    id_vars='category',  #category는 고정하기
    # var_name= 'month'
    value_name='amount'  #가value 영역에는 amount 매출
)
```
