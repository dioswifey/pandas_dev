```python
import pandas as pd     
```

```python
orders = pd.read_csv('../data/orders.csv', parse_dates=['order_datetime'])
```

```python
orders.info()
```

```python
# 1. NaN drop, order_datetime column
o = orders.dropna(subset=['order_datetime'])
o.info()
```

```python
#month 추출
# o.head()

o['month'] = o['order_datetime'].dt.month
o.head()
```

```python


jan = o[o['month'] == 1]
feb = o[o['month'] == 2]

print(f'1월 개수: {jan.shape[0]}, 2월 개수: {feb.shape[0]}')
```

```python
j_f = pd.concat([jan,feb],axis=0) #위아래 붙이기
# j_f = pd.concat([jan,feb],axis=1ㅡ 9) #좌우로 붙이기 <- error  jan과 feb 같은 값이여야지 옆으로 붙는데 안됨
print(f'1월-2월 주문 개수: {j_f.shape[0]}')
```
