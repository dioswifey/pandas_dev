```python
import pandas as pd
```

```python
# pivot
df = pd.DataFrame({
    'date' :['2026-01', '2026-01', '2026-02', '2026-02'],
    'city' : ['seoul', 'busan', 'seoul', 'busan'],
    'temp' : [-2.5, 1.0, 0.5, 3.5],
    'humid' : [40,55,45,60]
})
df
```

```python
#pivot()


wide = df.pivot(
    columns='city',
    index='date',
    values= 'temp'
)
wide
```

```python
#values를 multi 로 조정 가능(temp, humid)

result = df.pivot(  #result는 DataFrame[컬럼명]
    columns='city',
    index='date',
    values= ['temp', 'humid']
)

result

# result['humid']
```

```python
result['temp']
```

```python
result['temp', 'busan']
```

```python
#melt 
df = pd.DataFrame({
    'name' : ['홍길동', '고길동'],
    'kor': [90, 80],
    'eng' : [85, 85],
    'grade' : ['A', 'B']
})
df
```

```python
#melt()가 kor, eng 두 컬럼을 세로로 녹여서 다음 두 컬럼으로 재구성
df.melt(
    id_vars='name',
    value_vars=['kor', 'eng']  #values 에 들어갈 컬럼 지정    
)
```

```python
df.melt(
    id_vars=['name', 'grade'],
    value_vars = ['kor', 'eng'],
    var_name = 'subject',
    value_name = 'score'
)

# 이렇게 melt()한 뒤 다시 pivot()하는 경우는 보통:
# 분석하기 쉬운 long 형태로 바꿔서 작업한 뒤, 보기 좋은 wide 형태로 다시 펼칠 때
.
```

```python
#stack : 컬럼을 인덱스로

# stack()은 가로에 있는 컬럼 이름을 아래로 내려서 행 index로 만드는 메서드입니다
# 메소드 원형: DataFrame.stack(level=-1, dropna=<no_default>, sort=True, future_stack=True)   DataFrame | Series
# level 인덱스로 내릴 컬럼의 레벨, 정수, 레벨명(문자열), 리스트, -1 : 가장 안쪽(마지막)레벨
```

```python
stack_df = pd.DataFrame({
    'kor': [90, 80],
    'eng': [85, 85]
})

stack_df #DatdaFrame
# stack_df.columns
```

```python
stack_df.columns.name = 'subject'
stack_df
```

```python
stack_df.index.name='name'
stack_df
```

```python
print(stack_df)
print(stack_df.stack())   #Kor, eng (columns들)-> index로 보낸다   vertical 하게 stack하기 때문
```

```python
import numpy as np
```

```python
np.arange(8)  #arrange(start=0, end, step=1)
```

```python
#다중 컬럼: 특정 레벨
np.arange(8).shape   #1차원

np.arange(8).reshape(2,4) #2 by 4  <- 2차원으로 변경
# array([[0, 1, 2, 3],
#        [4, 5, 6, 7]])
```

```python
mi = pd.DataFrame(
    data=np.arange(8).reshape(2,4),
    index= ['r1', 'r2'],
    columns = pd.MultiIndex.from_product(
        [['A', 'B'], ['x', 'y']],
        names=['grp', 'sub']
    )
)
mi
```

```python
#stack에서 level 다루기
mi.stack(level='sub')
```

```python
mi.stack(level=['grp', 'sub'])
```
