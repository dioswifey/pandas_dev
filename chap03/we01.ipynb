```python
#행 기준 추출 : loc[행, 열] => 문자열, iloc[행, 열] =>숫자
#주문 데이터 읽기
orders_df = pd.read_csv(
    '../data/orders.csv',
    parse_dates=['order_datetime']    #order_datetime 컬럼(str)을 datetime64 변경
)
orders_df.info()
```

```python
#행 기준 추출: loc[행, 열] => 문자열, iloc[행, 열] => 숫자
# comma 가 없으면 column을 아예 안 선택한거임 -> 모든 column들이 나옴

orders_df.loc[0:5]      # 6행이나옴 - 모든 열 (모든 column들)  범위지정 (0이라는 값부터 5라는 값까지)
# orders_df.iloc[0:5]      # 5행이나옴  start 부터 end -1 까지
```

```python
# 행, 열 을 둘다 integer로 접근해서 특정 value 추출
print(orders_df.head())
print(orders_df.iloc[0,1])
```
