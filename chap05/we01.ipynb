```python
import pandas as pd
```

```python
#고객 데이터 로딩 (적재)
cust_df = pd.read_csv('../data/customers.csv')
```

```python
cust_df.info()
```

```python
#열별 결측 개수와 비율 진단

# 한번에 컬럼별 null의 개수를 셀 수 있음
# cust_df : DataFram.isna() => 컬럼 7ro
# cust_df : Dataframe
print(f'결측 개수 : {cust_df.isna().sum()}')
print(f'결측 비율 : {cust_df.isna().mean() * 100}')
```

```python
#missing value 처리
#원본을 보호
cleaned = cust_df.copy() #복사해서 대입, 원본 보호

#1. gender column: "미상" 채우기
cleaned['gender']=cleaned['gender'].fillna('미상')

#2. birth_date 값이 없는 행은 삭제 : dropna는 행 삭제 , 근데 null인 행을 삭제해야함
age_ready = cust_df.dropna(subset=['birth_date'],)
# age_ready = cust_df.dropna(subset=['birth_date'], inplace=True)   이건 원본에 반환
#3. email: 분석에 사용하지 않는다

print(f'gender 미상 채운 뒤 결측 : {cleaned["gender"].isna().sum()}')
print(f'birth_date 기준 dropna 전 행수 : {len(cust_df)}')
print(f'birth_date 기준 dropna 후 행수 : {len(age_ready)}')
```
