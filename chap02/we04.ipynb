```python
#missing data handling
# 결측 개수 
# 고객데이터
customers_df = pd.read_csv('../data/customers.csv')
customers_df.head()
```

```python
customers_df.info()
```

```python
# 결측치 - 생일 데이터 없는 고객의 수 확인
customers_df['birth_date'].isna()  #isna() 없니? 없으면 true, 있으면 false

```

```python
#단일 컬럼의 결측 개수
customers_df['birth_date'].isna().sum()   #없는 거 (true)의 개수
```

```python
#다중 컬럼의 결측 개수 세기
customers_df.isna().sum()
```
