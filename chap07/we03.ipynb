```python
import pandas as pd
```

```python
# 1) 고객데이터에서 나이를 연령대로 분리
cust = pd.read_csv('../data/customers.csv')
```

```python
cust['birth_date']
```

```python
cust.info()
```

```python
# 1.1 Null(NaN) 제거
cust.dropna(subset=['birth_date'])  #복사본으로 작업한 것이 된다
cust = cust.dropna(subset=['birth_date'])
cust.info()
```

```python
cust.info()
```

```python
#1.2 birth_date 데이터 타입 현재 "str"이다
#dtype을 datetime64로 변경
cust['birth_date']= pd.to_datetime(cust['birth_date'])
cust.info()
```

```python
# 1. 3 나이구하기    현재 연도(2026) - 생년
cust['age'] = 2026 - cust['birth_date'].dt.year
cust.head()
```

```python
# 1. 4 연령대 구간화  10대, 20대, 30대, 40대, 50 대, 60 대 이상
bins = [0, 19, 29, 39, 49, 59, 200]
labels = ['10대', '20대', '30대', '40대', '50대', '60대 이상']

cust['age_group'] = pd.cut(cust['age'], bins=bins, labels=labels)
```

```python
cust['age_group'].value_counts().sort_index()
```

```python
# 2) 가격을 균등 분할
# 데이터 로딩: 상품데이터
prod = pd.read_csv('../data/products.csv') 
prod.head()
```

```python
prod.info()
```

```python
#가격 데이터에 대한 정제
#2.1 원, , => str , 숫자 변경
price = (
    prod['price'].astype(str)
    .str.replace('원', '', regex = False)
    .str.replace(',', '', regex=False)
)

type(price) #dt: series   => numeric에 바로 넣을 수 있음
```

```python
#2.2 숫자 변경 : pd.to_numeric(Series, error)
prod['price'] =pd.to_numeric(price, errors='coerce')
```

```python
#2.3 가격이 0보다 큰 행 추출
# 0이거나 작은 행은 삭제
#위에서 numeric으로 바꿨기때문에 0보다 크다를 비교할수있다

prod = prod[prod['price']>0]
prod.head()
```

```python
# 분위수 기준 4등급 (같은  개수씩으로 분할 된다)   
prod['price_tier'] = pd.qcut(
    prod['price'], #분할할 데이터 지정
    q=4,           #분할할 개수
    labels=['저가', '중가', '고가', '최고가']
)

prod['price_tier'].head()
# 1    고가
# 2    저가
# 3    고가
# 4    저가
# 5    저가
```

```python
prod['price'].describe()
```

```python
prod['price_tier'].value_counts()
```

```python
# 3) status : 읽기 쉬운 라벨 매핑
# orders.csv

#주문 데이터 로딩
orders = pd.read_csv('../data/orders.csv')
```

```python
orders['status'].unique()
#canceled 나온걸 읽기쉬운 취소로 바꾸기   
```

```python
smap = {
    'paid':'결제완료',
    'delivered':'배송완료',
    'shipped':'배송중',
    'canceled':'취소',
    'returned':'반품'
}

orders['status_ko'] = orders['status'].map(smap)
orders['status_ko'].value_counts()
```
