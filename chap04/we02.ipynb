```python
import pandas as pd
```

```python
# 웹으로 주문된 주문중에 배송중인 주문 수
# 두개의 condition으로 분리하기

orders_df = pd.read_csv('../data/orders.csv', parse_dates=['order_datetime'])
orders_df.head()
```

```python
# 두개의 condition으로 분리하기
# 조건 2개를 con1 & con2  를 이런식으로 연결 할 수 있다

con1 = orders_df['channel'] == 'web'
con2 = orders_df['status'] == 'delivered'

orders_df[con1 & con2].shape[0]
```

```python
# not  하려면 ~

#웹이 아닌 경로로 들어온거로 boolean
[~(orders_df['channel'] == 'web')]
```

```python
#'app', 'canceled' 주문수
con1 = orders_df['channel'] == 'app'
con2 = orders_df['status'] == 'canceled'

orders_df[con1 & con2].shape[0]
```
